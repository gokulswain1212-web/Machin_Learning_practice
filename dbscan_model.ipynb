```python
# import libraries for DBSCAN 
import pandas as pd
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
# load the dataset
df=pd.read_csv(r"C:\Users\LENOVO\Desktop\Machine learning\DBSCAN_model\dbscan_similar_2d_dataset.csv")
df
```

```python
# remove true_group
df.drop('true_group',axis=1,inplace=True)
```

```python
# plot the dataset
plt.scatter(df['x'],df['y'])
plt.title('Dataset')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# create a DBSCAN model
dbscan_model = DBSCAN(eps=2, min_samples=10)
dbscan_model.fit(df)
# display the labels assigned to each data point
labels = dbscan_model.labels_
labels
```

```python
# make the plot using the labels assigned to each data point
plt.scatter(df['x'], df['y'], c=labels, cmap='rainbow')
plt.title('DBSCAN Clustering')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# check silhouette score
from sklearn.metrics import silhouette_score

silhouette_avg = silhouette_score(df, labels)
print(f"Average silhouette score: {silhouette_avg}")
```
